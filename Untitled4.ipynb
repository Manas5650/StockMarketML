```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("all_stocks_5yr.csv",parse_dates=['date'])
```

```python
print("First 5 rows:\n", df.head(),"\n")
print("Info:\n")
print("\nMissing values:\n",df.isna().sum())
```

```python
df.dropna(subset=['close'],inplace=True)
df.sort_values('date', inplace=True)
```

```python
stock_name = "AAPL"

df_stock = df[df['Name'] == stock_name]

plt.figure(figsize=(12,6))
sns.lineplot(data=df_stock, x='date', y='close')
plt.title(f'{stock_name} Close Price Over Time')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df_stock['close'], bins=30, kde=True)
plt.title(f'{stock_name} Price Distribution')
plt.xlabel('Close Price')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```
